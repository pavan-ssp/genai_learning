```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
FILE_PATH=r"C:\Users\july14\llama2.pdf"
```

```python
loader=PyPDFLoader(FILE_PATH)
pages=loader.load()
```

```python
pages = []
async for page in loader.alazy_load():
    pages.append(page)

print(pages[0])    
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,#hyperparameter
    chunk_overlap=50 #hyperparemeter
)

split_docs = splitter.split_documents(pages)
split_docs
```

```python
len(split_docs)
```

```python
import faiss
from langchain_community.vectorstores import FAISS
from langchain_community.docstore.in_memory import InMemoryDocstore
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings=HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
from langchain_community.vectorstores import FAISS
from langchain_community.docstore.in_memory import InMemoryDocstore
import faiss

# Define index (for 384-dimensional embeddings)
index = faiss.IndexFlatIP(384)

# Example empty docstore
docstore = InMemoryDocstore({})

# Assuming 'embeddings' is an embedding function already defined
vector_store = FAISS(
    embedding_function=embeddings,
    index=index,
    docstore=docstore,
    index_to_docstore_id={}
)
```

```python
vector_store.add_documents(documents=split_docs)
```

```python
retriever=vector_store.as_retriever(
    search_kwargs={"k": 10} #hyperparameter
)
```

```python
retriever.invoke("what is llama model?")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
model=ChatGoogleGenerativeAI(model='gemini-1.5-flash')
```

```python
from langchain import hub
prompt = hub.pull("rlm/rag-prompt")
```

```python
import pprint
pprint.pprint(prompt.messages)
```

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
```

```python
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)
    
```

```python
rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | model
    | StrOutputParser()
)
```

```python
rag_chain.invoke("what is llama model?")
```
